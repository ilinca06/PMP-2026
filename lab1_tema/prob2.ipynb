```python
import numpy as np
import matplotlib.pyplot as plt

def joaca_joc(p_stema):
    suma = 0 
    
    while True:
        # Alegem 1 (stema) sau 0 (ban) 
        moneda = np.random.choice([1, 0], p=[p_stema, 1 - p_stema])
        
        if moneda == 1:
            z = np.random.randint(1, 7) # Zarul ia valori intre 1 si 6 
            suma += (z - 3)
            break
        else:
            # Primul jucator pierde 0.5, deci datoria celui de-al doilea scade cu 0.5
            suma -= 0.5 
            
    return suma

def simuleaza(numar_jocuri, p_stema):
    # Generare lista cu sumele finale
    sume_totale = [joaca_joc(p_stema) for _ in range(numar_jocuri)]
        
    print(f"Media sumei S pentru p={p_stema} este: {np.mean(sume_totale):.2f} $")
    
    # Afisare grafic (histograma)
    plt.hist(sume_totale, bins=15, alpha=0.7, edgecolor='black')
    plt.title(f'Distributia sumei S (probabilitate stema = {p_stema})')
    plt.xlabel('Suma totala S ($)')
    plt.ylabel('Frecventa')
    plt.show()

# Punctul c) - moneda corecta
simuleaza(10000, 0.5)

# Punctul d) - moneda masluita
simuleaza(10000, 0.3)
simuleaza(10000, 0.7)
```
