```python
import numpy as np
import matplotlib.pyplot as plt

numar_clienti = 10000
timpi_servire = []

# Distribuim clientii la cei 3 frizeri folosind probabilitatile cerute
frizeri_alesi = np.random.choice([1, 2, 3], size=numar_clienti, p=[3/13, 6/13, 4/13])

for frizer in frizeri_alesi:
    # Parametrul 'scale' reprezinta 1/lambda pentru distributia exponentiala
    if frizer == 1:
        timp = np.random.exponential(scale=1/3)
    elif frizer == 2:
        timp = np.random.exponential(scale=1/6)
    else:
        timp = np.random.exponential(scale=1/4)
    timpi_servire.append(timp)

timpi_servire = np.array(timpi_servire)

print(f"Media estimata a timpului de servire (X): {np.mean(timpi_servire):.4f} ore")
print(f"Deviatia standard estimata a lui X: {np.std(timpi_servire):.4f} ore")

# Vizualizare densitatea aproximativa a distributiei folosind o histograma de densitate
plt.hist(timpi_servire, bins=50, density=True, alpha=0.7, color='skyblue', edgecolor='black')
plt.title('Densitatea aproximativa a timpului de servire X')
plt.xlabel('Timp (ore)')
plt.ylabel('Densitate')
plt.show()
```
