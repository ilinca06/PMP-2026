```python
import csv
import random

# Creare fisier 
studenti_fictivi = [["Popescu Ion"], ["Ionescu Maria"], ["Radu Andrei"], ["Marin Elena"], ["Vasilescu George"]]

with open('studenti.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerows(studenti_fictivi)


def extrage_studenti(nume_fisier, k):
    studenti = []
    
    with open(nume_fisier, mode='r', encoding='utf-8') as fisier:
        cititor = csv.reader(fisier)
        for rand in cititor:
            if rand:
                studenti.append(rand[0])
                
    # Functia sample extrage k elemente unice, fara repetitie
    return random.sample(studenti, k)

# Rulare functie
alesi = extrage_studenti('studenti.csv', 3)
print("Studentii extrasi pentru tema sunt:", alesi)
```
