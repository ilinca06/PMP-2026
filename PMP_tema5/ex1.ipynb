# Exercițiul 1: Probabilitatea unei secvențe de observații (Forward)

Conform definiției, variabila Forward este probabilitatea comună a observațiilor de până la momentul $t$ și a stării curente: 
$\alpha_t(j) = P(O_1=o_1, ..., O_t=o_t, X_t=j)$.

### a) Inițializare
Pentru primul moment de timp ($t=1$), formula devine $\alpha_1(j) = P(O_1=o_1, X_1=j)$.
Aplicând regula produsului (definiția probabilității condiționate), putem rescrie această probabilitate comună astfel:
$$P(O_1=o_1, X_1=j) = P(O_1=o_1 | X_1=j) \cdot P(X_1=j)$$
Din definițiile modelului HMM:
* $P(X_1=j)$ reprezintă distribuția inițială a stărilor, adică $\pi_j$.
* $P(O_1=o_1 | X_1=j)$ reprezintă probabilitatea de emisie, adică $b_j(o_1)$. 

Prin urmare, înlocuind termenii, obținem: **$\alpha_1(j) = \pi_j \cdot b_j(o_1)$**.

### b) Recurență
Pornim de la definiția pentru momentul $t$ și aplicăm legea probabilității totale, condiționând după starea în care s-a aflat sistemul la pasul anterior ($X_{t-1} = i$):
$$\alpha_t(j) = P(O_{1:t}, X_t=j) = \sum_i P(O_{1:t}, X_t=j, X_{t-1}=i)$$

Folosind regula înlănțuirii (chain rule) pentru probabilități comune, descompunem termenul din interiorul sumei:
$$= \sum_i P(O_t | X_t=j, X_{t-1}=i, O_{1:t-1}) \cdot P(X_t=j | X_{t-1}=i, O_{1:t-1}) \cdot P(X_{t-1}=i, O_{1:t-1})$$

Acum aplicăm cele două ipoteze fundamentale ale unui Lanț Markov Ascuns (HMM):
1. **Ipoteza de observare:** Observația curentă $O_t$ depinde doar de starea curentă $X_t$. Deci: $P(O_t | X_t=j, \dots) = b_j(o_t)$.
2. **Ipoteza Markov:** Starea curentă $X_t$ depinde doar de starea anterioară $X_{t-1}$. Deci: $P(X_t=j | X_{t-1}=i, \dots) = a_{ij}$.

Observăm că ultimul termen din înmulțire este exact definiția variabilei Forward de la pasul $t-1$: 
$$P(X_{t-1}=i, O_{1:t-1}) = \alpha_{t-1}(i)$$

Înlocuind aceste simplificări în suma noastră, obținem:
$$\alpha_t(j) = \sum_i b_j(o_t) \cdot a_{ij} \cdot \alpha_{t-1}(i)$$
Deoarece probabilitatea de emisie $b_j(o_t)$ nu depinde de indicele sumei $i$, o putem scoate ca factor comun în fața acesteia, rezultând demonstrația cerută:
**$$\alpha_t(j) = b_j(o_t) \sum_i \alpha_{t-1}(i) a_{ij}$$**

### c) Finalizare
Scopul final al algoritmului Forward este să calculeze probabilitatea întregii secvențe observate: $P(O_1=o_1, ..., O_T=o_T)$.
Aplicăm din nou legea probabilității totale, de data aceasta sumând peste toate stările posibile în care s-ar putea afla sistemul la ultimul moment de timp $T$:
$$P(O_{1:T}) = \sum_j P(O_{1:T}, X_T=j)$$
Termenul din interiorul sumei este chiar definiția variabilei Forward la momentul final $T$. Așadar, probabilitatea totală a secvenței este suma valorilor Forward finale:
**$$P(O_{1:T}) = \sum_j \alpha_T(j)$$**
